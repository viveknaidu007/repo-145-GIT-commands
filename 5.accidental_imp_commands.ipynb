### when we accidentally , commit the chnages to brnach ,but not done synced ,because u are not in actual brnach name , then we can revert them back to back as before to staged coomits 

```python
git reset --soft HEAD~1
```

```python
What this does:

    --soft: This is the most important part. It undoes the commit but keeps all your code changes as "staged" (ready to be committed again). You won't lose a single line of code.

    HEAD~1: This tells Git to go back by one commit. If you made three commits that you want to undo, you would use HEAD~3.
```

# ********* merge conflicts **********

Here’s a quick reference of useful **Git commands for merge handling**, especially when dealing with **conflicts**, **abort**, or **resetting**:

---

## 🔄 **Merge Commands**

### ✅ 1. Start a merge

```bash
git merge <branch-name>
```

Merges the specified branch into your current branch.

---

## ❌ **Abort Merge**

### ✅ 2. Abort a merge in progress (if there are conflicts):

```bash
git merge --abort
```

> Restores your working tree to the state **before the merge** started.

⚠️ Works only if you're in the middle of a merge (after `git merge` and before resolving conflicts).

---

## 🧹 **Reset or Undo**

### ✅ 3. Reset to a previous commit (discard all changes):

```bash
git reset --hard <commit-hash>
```

### ✅ 4. Reset a branch to remote state:

```bash
git fetch origin
git reset --hard origin/<branch-name>
```

---

## 🔁 **Revert a Merge Commit**

### ✅ 5. Revert a completed merge (non-destructive, creates a new commit):

```bash
git revert -m 1 <merge-commit-hash>
```

* `-m 1` means "keep parent 1" (your branch before the merge).
* Safe to use even after pushing to remote.

---

## ⚙️ **Conflict Resolution Commands**

### ✅ 6. See merge conflicts:

```bash
git status
```

### ✅ 7. Mark conflicts as resolved:

```bash
git add <resolved-files>
git commit
```

---

## 🧪 Miscellaneous

### ✅ 8. See all branches merged into current:

```bash
git branch --merged
```

### ✅ 9. See merge commit history:

```bash
git log --merges
```

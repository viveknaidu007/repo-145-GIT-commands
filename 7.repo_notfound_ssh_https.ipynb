## 1️⃣ SSH error: `Permission denied (publickey)`

This happens when your system is trying to use SSH (`git@github.com:...`) but **your SSH key is not set up or not linked to GitHub**.

### Fix it:

### Step 1: Check if you already have an SSH key

```bash
ls ~/.ssh
```

Look for files like `id_rsa` or `id_ed25519`.

---

### Step 2: If not, generate one

```bash
ssh-keygen -t ed25519 -C "your_email@example.com"
```

---

### Step 3: Add SSH key to agent

```bash
eval "$(ssh-agent -s)"
ssh-add ~/.ssh/id_ed25519
```

---

### Step 4: Copy and add to GitHub

```bash
cat ~/.ssh/id_ed25519.pub
```

Now go to:
👉 GitHub SSH Keys Settings
Add the key there.

---

### Step 5: Test connection

```bash
ssh -T git@github.com
```

If it works, retry:

```bash
git clone git@github.com:ByteCraft-Labs/SignatureVerification.git
```

---

## 2️⃣ HTTPS error: `Repository not found`

This is a **different problem**. It usually means:

* ❌ The repo **doesn’t exist**
* 🔒 The repo is **private**
* 🔗 The URL is **wrong**

---

### Check this first:

Open in browser:
👉 [https://github.com/ByteCraft-Labs/SignatureVerification](https://github.com/ByteCraft-Labs/SignatureVerification)

If you see:

* **404** → repo is deleted or renamed
* **Private repo** → you need access

---

## 🔥 Most likely situation

Given both errors together:

👉 The repository is probably **private or no longer exists**

---

## ✅ Solutions depending on situation

### ✔️ If repo is private

* Ask owner to give you access
* Then use:

```bash
git clone https://<your-username>@github.com/ByteCraft-Labs/SignatureVerification.git
```

---

### ✔️ If repo was renamed

Find correct repo URL and clone again

---

### ✔️ If you just need similar project

Search on GitHub:

```
Signature Verification Python
```

---

## 💡 Quick workaround (recommended)

If you’re not specifically required to use SSH:

👉 Use HTTPS **with correct repo URL**

```bash
git clone https://github.com/<correct-owner>/<correct-repo>.git
```

---

## ⚠️ Key takeaway

* SSH error = **auth problem**
* HTTPS "not found" = **repo problem**

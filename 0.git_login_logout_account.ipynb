Once you run `gh auth login`, the GitHub CLI will guide you through a series of prompts. Here is exactly what to select and the commands to manage your accounts afterward.

---

### 1. How to complete `gh auth login`
After typing the command, follow these selections in your terminal:

1.  **What account do you want to log into?** * Select: `GitHub.com`
2.  **What is your preferred protocol for Git operations?** * Select: `HTTPS`
3.  **Authenticate Git with your GitHub credentials?** * Select: `Yes`
4.  **How would you like to authenticate GitHub CLI?** * Select: `Login with a web browser`
5.  **Copy the One-Time Code:** * The terminal will show an 8-character code (e.g., `1234-ABCD`).
    * Press **Enter**. This will open your default browser.
    * Paste the code into the browser and click **Authorize github**.

---

### 2. How to see all logged-in accounts
To see which account is currently active and check for any other saved credentials, use:

```powershell
gh auth status
```
* **What it shows:** It will list your username, the authentication method (token or browser), and your active scopes (like `repo`, `gist`, etc.).

---

### 3. How to logout of a GitHub account
If you want to remove an account from your machine:

```powershell
gh auth logout
```
* **Note:** If you have multiple accounts logged in, it will ask you which one you want to log out of. 
* If you want to force a logout for a specific host, use: `gh auth logout -h github.com`

---

### 4. How to switch between accounts
If you work with multiple accounts (e.g., a personal one and a work one), you don’t need to log out every time. You can use:

```powershell
gh auth switch
```
* This will show a list of all logged-in users. Use the arrow keys to select the one you want to make "Active."

### Summary Table of Commands

| Action | Command |
| :--- | :--- |
| **Login** | `gh auth login` |
| **Check Status** | `gh auth status` |
| **Switch Users** | `gh auth switch` |
| **Logout** | `gh auth logout` |

Once you finish the `gh auth login` process, your `git clone` command from earlier should work perfectly without asking for a password.

If a repository is **private**, GitHub will return a `404 Not Found` error instead of a `403 Forbidden` for security reasons. This prevents people from even knowing a private project exists unless they are authenticated.

Here are the three best ways to fix this in your Windows terminal:

---

### 1. The GitHub CLI Method (Recommended)
This is the easiest way to handle authentication on Windows. If you don't have it, you can install it via [cli.github.com](https://cli.github.com/) or just run `winget install --id GitHub.cli` in your terminal.

1.  Run the login command:
    ```bash
    gh auth login
    ```
2.  Follow the prompts: Select **GitHub.com** > **HTTPS** > **Yes** (to authenticate with your browser).
3.  Once the browser window opens, click **Authorize**.
4.  Try cloning again:
    ```bash
    git clone https://github.com/aivivekai/signature-detect.git
    ```

---

### 2. Using a Personal Access Token (PAT)
GitHub no longer accepts your account password for terminal commands. You need a **Token**.

1.  Go to **GitHub Settings** > **Developer Settings** > **Personal Access Tokens** > **Tokens (classic)**.
2.  Generate a new token with the `repo` scope selected.
3.  **Copy the token immediately** (you won't see it again).
4.  When you clone, use this format in your terminal:
    ```bash
    git clone https://<your_token>@github.com/aivivekai/signature-detect.git
    ```
    *Replace `<your_token>` with the string GitHub gave you.*

---

### 3. Clear the Git Credential Manager
Sometimes Windows stores old, "stale" credentials that block the new ones.

1.  Open the **Start Menu** and search for **Credential Manager**.
2.  Go to **Windows Credentials**.
3.  Find any entries for `git:https://github.com` and **Remove** them.
4.  The next time you run `git clone`, a popup should appear asking you to sign in via your browser.

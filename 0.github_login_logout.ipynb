If a repository is **private**, GitHub will return a `404 Not Found` error instead of a `403 Forbidden` for security reasons. This prevents people from even knowing a private project exists unless they are authenticated.

Here are the three best ways to fix this in your Windows terminal:

---

### 1. The GitHub CLI Method (Recommended)
This is the easiest way to handle authentication on Windows. If you don't have it, you can install it via [cli.github.com](https://cli.github.com/) or just run `winget install --id GitHub.cli` in your terminal.

1.  Run the login command:
    ```bash
    gh auth login
    ```
2.  Follow the prompts: Select **GitHub.com** > **HTTPS** > **Yes** (to authenticate with your browser).
3.  Once the browser window opens, click **Authorize**.
4.  Try cloning again:
    ```bash
    git clone <>
    ```

---

### 2. Using a Personal Access Token (PAT)
GitHub no longer accepts your account password for terminal commands. You need a **Token**.

1.  Go to **GitHub Settings** > **Developer Settings** > **Personal Access Tokens** > **Tokens (classic)**.
2.  Generate a new token with the `repo` scope selected.
3.  **Copy the token immediately** (you won't see it again).
4.  When you clone, use this format in your terminal:
    ```bash
    git clone https://<your_token>@github.com/aivivekai/signature-detect.git
    ```
    *Replace `<your_token>` with the string GitHub gave you.*

---

### 3. Clear the Git Credential Manager
Sometimes Windows stores old, "stale" credentials that block the new ones.

1.  Open the **Start Menu** and search for **Credential Manager**.
2.  Go to **Windows Credentials**.
3.  Find any entries for `git:https://github.com` and **Remove** them.
4.  The next time you run `git clone`, a popup should appear asking you to sign in via your browser.

---

# 🔍 1. Check which Git account is configured

```bash
git config --global user.name
git config --global user.email
```

👉 This shows the **global Git identity** (used for commits).

If you want to check **local repo config**:

```bash
git config user.name
git config user.email
```

---

# 🌐 2. Check which GitHub account is authenticated

Git itself doesn’t store “logged-in user” like a session. Authentication is handled by credentials.

### If using GitHub CLI (recommended):

```bash
gh auth status
```

---

# 🚪 3. Logout from GitHub account

### Using GitHub CLI:

```bash
gh auth logout
```

---

### If using credential manager (Windows):

```bash
git credential-manager reject https://github.com
```

OR manually remove:

```bash
cmdkey /delete:git:https://github.com
```

---

# 🔄 4. Switch to another GitHub account

### Step 1: Remove old credentials

```bash
cmdkey /delete:git:https://github.com
```

---

### Step 2: Do a Git operation (it will ask login)

```bash
git pull
```

👉 Enter:

* New GitHub username
* Personal Access Token (NOT password)

---

# 👤 5. Change Git commit identity (important)

```bash
git config --global user.name "Your Name"
git config --global user.email "your-email@example.com"
```

---

# 🔁 6. (Optional) Use different account per repo

```bash
git config user.name "Other Name"
git config user.email "other-email@example.com"
```

# **************

Managing multiple GitHub accounts safely is mostly about **avoiding credential conflicts** and making sure each repo uses the correct identity. The cleanest approach is using **SSH keys + per-repo config**.

---

# 🔐 Best Practice: Use SSH for Multiple Accounts

## 1. Generate separate SSH keys

Open CMD:

```bash
ssh-keygen -t ed25519 -C "your-email1@example.com"
```

Save as:

```
C:\Users\YourUser\.ssh\id_ed25519_personal
```

Then for second account:

```bash
ssh-keygen -t ed25519 -C "your-email2@example.com"
```

Save as:

```
C:\Users\YourUser\.ssh\id_ed25519_work
```

---

## 2. Add keys to SSH agent

```bash
ssh-add ~/.ssh/id_ed25519_personal
ssh-add ~/.ssh/id_ed25519_work
```

---

## 3. Add SSH keys to GitHub accounts

Copy keys:

```bash
type ~/.ssh/id_ed25519_personal.pub
```

```bash
type ~/.ssh/id_ed25519_work.pub
```

👉 Add each key in:
GitHub → Settings → SSH and GPG keys

---

## 4. Configure SSH config file

Edit:

```
C:\Users\YourUser\.ssh\config
```

Add:

```bash
# Personal account
Host github-personal
    HostName github.com
    User git
    IdentityFile ~/.ssh/id_ed25519_personal

# Work account
Host github-work
    HostName github.com
    User git
    IdentityFile ~/.ssh/id_ed25519_work
```

---

## 5. Clone repos using specific account

### Personal repo:

```bash
git clone git@github-personal:username/repo.git
```

### Work repo:

```bash
git clone git@github-work:company/repo.git
```

---

## 6. Set correct identity per repo

Inside each repo:

```bash
git config user.name "Your Name"
git config user.email "your-email@example.com"
```

---

# 🔄 Switching accounts (simple workflow)

* No logout needed
* Just clone/use correct SSH host (`github-personal` or `github-work`)
* Git automatically uses the right key

---

# ⚠️ Common mistakes to avoid

* ❌ Using HTTPS for multiple accounts → causes credential conflicts
* ❌ Using same SSH key for both accounts
* ❌ Forgetting to set repo-specific email

---

# *************************************

Here’s a **clean, step-by-step setup (Windows CMD)** to use **two GitHub accounts (work + personal) with SSH** — this avoids constant login/logout headaches.

---

# 🔐 1. Generate two SSH keys

### 👉 Personal account

```bash
ssh-keygen -t ed25519 -C "your-personal-email@example.com"
```

When asked for file name:

```
C:\Users\YourUser\.ssh\id_ed25519_personal
```

---

### 👉 Work account

```bash
ssh-keygen -t ed25519 -C "your-work-email@example.com"
```

Save as:

```
C:\Users\YourUser\.ssh\id_ed25519_work
```

---

# 🔑 2. Start SSH agent & add keys

```bash
ssh-agent -s
ssh-add ~/.ssh/id_ed25519_personal
ssh-add ~/.ssh/id_ed25519_work
```

---

# 📋 3. Copy public keys

```bash
type %USERPROFILE%\.ssh\id_ed25519_personal.pub
```

```bash
type %USERPROFILE%\.ssh\id_ed25519_work.pub
```

---

# 🌐 4. Add keys to GitHub

Go to:
👉 GitHub → Settings → **SSH and GPG keys**

* Add **personal key** to personal account
* Add **work key** to work account

---

# ⚙️ 5. Create SSH config file

Open (create if not exists):

```
C:\Users\YourUser\.ssh\config
```

Add:

```bash
# Personal GitHub
Host github-personal
    HostName github.com
    User git
    IdentityFile ~/.ssh/id_ed25519_personal

# Work GitHub
Host github-work
    HostName github.com
    User git
    IdentityFile ~/.ssh/id_ed25519_work
```

---

# 🔗 6. Clone repos correctly

### Personal repo:

```bash
git clone git@github-personal:your-username/repo.git
```

### Work repo:

```bash
git clone git@github-work:company-username/repo.git
```

---

# 👤 7. Set correct identity per repo

Inside each repo:

### Personal:

```bash
git config user.name "Your Name"
git config user.email "your-personal-email@example.com"
```

### Work:

```bash
git config user.name "Your Work Name"
git config user.email "your-work-email@example.com"
```

---

# ✅ 8. Test connections

```bash
ssh -T git@github-personal
```

```bash
ssh -T git@github-work
```

You should see:

```
Hi username! You've successfully authenticated...
```

---

# ⚠️ Important tips

* Always use `github-personal` or `github-work` (not `github.com`)
* Each repo sticks to one account automatically
* No need to logout/switch manually anymore

---

# 🚀 Result

* Both accounts work **side-by-side**
* No credential conflicts
* Clean, professional setup
